```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
x = np.array([
    [1, 8],
    [2, 13],
    [3, 22],
    [4, 27],
    [5, 36]
])
```

```python
y = np.array([40, 48, 57, 65, 74])
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_scaled, y, test_size = 0.4, random_state = 42)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)

y_pred = model.predict(x_test)
```

```python
from sklearn.linear_model import RidgeCV
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

model = make_pipeline(
    StandardScaler(),
    RidgeCV(alphas = [0.01, 0.1, 1, 10, 100])
)

model.fit(x_train, y_train)

ridge_model = model.named_steps["ridgecv"]
print("Best alpha : ", ridge_model.alpha_)
```

```python
#Instead of lambda is alpha

model = make_pipeline(
    StandardScaler(),
    Ridge(alpha = 0.1)
    )

model.fit(x_train, y_train)
y_pred = model.predict(x_test)
```

```python
ridge_step = model.named_steps['ridge']

print("Intercept : ", ridge_step.intercept_)
print("Standard coefficients : ", ridge_step.coef_)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score

print("MAE       : ", mean_absolute_error(y_test, y_pred))
print("MSE       : ", mean_squared_error(y_test, y_pred))
print("RMSE      : ", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R-squared : ", r2_score(y_test, y_pred))
```
