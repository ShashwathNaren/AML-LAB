Feature_Selection : Variance Threshold

```python
import pandas as pd
import numpy as np
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.preprocessing import MinMaxScaler
from sklearn.feature_selection import VarianceThreshold
```

```python
data = pd.read_csv(r"C:\Users\bda\Desktop\261100660031\AML_LAB\Lab_8\data-1.csv")
```

```python
data.head()
```

```python
selector = VarianceThreshold(threshold = 0.0)
x_selected = selector.fit_transform(data)

selected_columns = data.columns[selector.get_support()]

print("Selected features : ", list(selected_columns))
```

```python
selector = VarianceThreshold(threshold = 0.1)
x_selected = selector.fit_transform(data)

selected_columns = data.columns[selector.get_support()]

print("Selected features : ", list(selected_columns))
```
