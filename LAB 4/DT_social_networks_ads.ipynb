```python
x=df.iloc[:, [2, 3]].values
x
```

```python
#spilt dependent variable
y=df.iloc[:, 4].values
y
```

```python
#split the data into training and testing
#test_size=0.2 -> training 80%, testing %20
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=0)
#X_train,X_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
X_test
#X_test.astype(int)
```

```python
from sklearn.preprocessing import StandardScaler
sc_X = StandardScaler()
X_train = sc_X.fit_transform(X_train)
X_test = sc_X.transform(X_test)
X_test
```

```python
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion= 'entropy', random_state=0)
#classifier = DecisionTreeClassifier(criterion= 'gini', random_state=0)
classifier.fit(X_train, y_train)
```

```python
y_pred = classifier.predict(X_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
Accuracy = (cm[0][0]+cm[1][1])/ (cm[0][0]+cm[0][1]+cm[1][0]+cm[1][1])
Accuracy
```

```python
from sklearn import metrics
#Model accuracy how often is classifier correct
prediction = metrics.accuracy_score(y_test, y_pred)
print("Accuracy : ", prediction*100, '%')
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
#predict for a new individual
new_individual = np.array([[29, 80000]])

new_individual_scaled = sc_X.transform(new_individual)

prediction = classifier.predict(new_individual_scaled)

if prediction[0] == 1:
    print("The individual is ready to purchase.")
else:
    print("The individual is predicted not to purchase")
```
