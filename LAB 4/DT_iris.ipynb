```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd 
```

```python
df = pd.read_csv(r"C:\Users\bda\Desktop\261100660031\iris.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
x = df.iloc[:,1:-1].values
x
```

```python
y = df.iloc[:, -1].values
y
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.25, random_state = 0)
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
x_test
```

```python
from sklearn.tree import DecisionTreeClassifier

classifier = DecisionTreeClassifier( criterion = 'entropy', random_state = 0)

classifier.fit(x_train, y_train)
```

```python
y_pred = classifier.predict(x_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)
cm
```

```python
Accuracy = (cm[0][0] + cm[1][1]) / (cm[0][0]+cm[0][1]+cm[1][0]+cm[1][1])
Accuracy
```

```python
from sklearn import metrics

prediction = metrics.accuracy_score(y_test, y_pred)
print("prediction : ", prediction*100, '%')
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
new = np.array([[4.2, 3.3, 1.2, 1]])
exist = np.array([[7.1,3,5.9,2.1]])

new_scaled = sc.transform(new)
exist_scaled = sc.transform(exist)

pred1 = classifier.predict(new_scaled)
pred2 = classifier.predict(exist_scaled)

print("For new : ")
if pred1 == 0:
    print("Iris-Setosa")
elif pred1 == 1:
    print("Iris-Versicolor")
elif pred1 == 2:
    print("Iris-Virginica")
else :
    print("Incorrectly classified")
    
print("For existed :")
if pred2 == 0:
    print("Iris-Setosa")
elif pred2 == 1:
    print("Iris-Versicolor")
elif pred2 == 2:
    print("Iris-Virginica")
else :
    print("Incorrectly classified")
```
